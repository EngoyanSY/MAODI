<h3>Преобразование в комплексные числа</h3>

```python
def contour_direction_vector(contours):
    complex_contours = []

    for cnt in contours:
        # приводим контур к виду (N,2)
        pts = cnt.reshape(-1, 2)

        # преобразуем в комплексные числа x + i*y
        z = pts[:, 0] + 1j * pts[:, 1]

        # разности между соседними точками (смещения)
        dz = np.diff(z)

        # заменим любые смещения >1 на sign(), т.к. шаг по пикселям
        dz = np.sign(dz.real) + 1j * np.sign(dz.imag)
        complex_contours.append(dz)

    return complex_contours
```

```python
def normalize_chain_code(dz_seq, target_length=64):
    if len(dz_seq) == 0:
        return np.zeros(target_length, dtype=complex)
    
    n = len(dz_seq)
    if n != target_length:
        # Индексы для равномерного отбора
        indices = np.linspace(0, n, target_length, endpoint=False)
        # Округляем до ближайших целых
        indices = np.round(indices).astype(int) % n
        dz_seq = dz_seq[indices]
    
    return dz_seq
```

```python
def scalar_product(z1, z2):
    n = int((len(z1) + len(z1)) / 2)
    z1_n = normalize_chain_code(z1, target_length=n)
    z2_n = normalize_chain_code(z2, target_length=n)
    return np.sum(z1_n*np.conj(z2_n))

```

```python
def VKF(z1, z2):
    VKF = []
    len_z2 = len(z2)
    for m in len_z2:
        VKF.append(scalar_product(z1, np.roll(z2, m)))
```

```python
def norma(z):
    result = 0
    for item in z:
        result = result + abs(item) ** 2
    return np.sqrt(result)
```

```python
def find_symbol(refer, z, symbol_by_index = None):
    t = []
    count = 0
    for cnt in refer:
        VKF = []
        len_z = len(z)
        for m in range(len_z):
            VKF.append(scalar_product(cnt, np.roll(z, m)))
        t_max = abs(max(VKF) / (norma(cnt) * norma(z)))
        # print(count, symbol_by_index[count], t_max)
        count = count + 1
        t.append(t_max)
    max_index = np.argmax(t) 
    return symbol_by_index[max_index]
```
