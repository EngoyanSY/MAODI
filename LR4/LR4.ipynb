<h3>Преобразование в комплексные числа</h3>

```python
def norm_vec_eq(z):
        return sum(abs(item) for item in z)
```

```python
def equalization(vector, p=128):
    if len(vector) < 2:
        return [0j] * p
    

    total_len = norm_vec_eq(vector)
    eq_len = total_len / p
    eq_vec = []
    
    vec_ost = vector[0]
    j = 0
    
    for _ in range(p):
        needed_len = eq_len
        current_sum = 0j
        
        while abs(vec_ost) < needed_len and j < len(vector) - 1:
            needed_len -= abs(vec_ost)
            current_sum += vec_ost
            j += 1
            vec_ost = vector[j]
            
        ratio = needed_len / abs(vec_ost) if abs(vec_ost) != 0 else 0
        part = vec_ost * ratio
        current_sum += part
        vec_ost -= part
        eq_vec.append(current_sum)
        
    return np.array(eq_vec)
```

```python
def normalize_contour(complex_vec):
    """Нормировка вектора: деление на норму для сопоставимости масштабов."""
    norm = np.sqrt(np.sum(np.abs(complex_vec)**2))
    if norm == 0: return complex_vec
    return complex_vec / norm
```

```python
def get_complex_vectors(contours, p=128):
    """Превращает список контуров OpenCV в список нормированных эквидистантных векторов."""
    complex_vectors = []
    for cnt in contours:
        pts = cnt.reshape(-1, 2)
        # 1. В комплексные числа (x + iy)
        z = pts[:, 0] + 1j * pts[:, 1]
        # 2. Векторы смещения (dz)
        dz = np.diff(z, append=z[0]) 
        # 3. Эквидистантность
        dz_eq = equalization(dz, p)
        # 4. Нормировка
        dz_norm = normalize_contour(dz_eq)
        complex_vectors.append(dz_norm)
    return complex_vectors
```

```python
def find_max_correlation_with_angle(z1, z2):
    """
    Вычисляет максимальную корреляцию и угол поворота между двумя контурами.
    """
    max_corr_val = -1
    best_angle = 0
    
    for m in range(len(z2)):
        corr = np.vdot(z1, np.roll(z2, m))
        if np.abs(corr) > max_corr_val:
            max_corr_val = np.abs(corr)
            best_angle = np.angle(corr) 
            
    return max_corr_val, np.degrees(best_angle)
```

```python
complex_refer = get_complex_vectors(clean_contours, p=128)
complex_target = get_complex_vectors(sorted_contours_target, p=128)
```

```python
results_list = []

print("Результаты идентификации:")
for i, target_vec in enumerate(complex_target):
    best_match = -1
    max_val = -1
    
    cnt = sorted_contours_target[i]
    x, y, w, h = cv2.boundingRect(cnt)
    M = cv2.moments(cnt)
    cy = int(M['m01'] / M['m00']) if M['m00'] != 0 else 0
    
    for ref_idx, ref_vec in enumerate(complex_refer):
        current_corr, angle = find_max_correlation_with_angle(ref_vec, target_vec)

        if abs(angle) > 45:
            current_corr *= 0.5
            
        if current_corr > max_val:
            max_val = current_corr
            best_match = ref_idx
            
    symbol = symbol_by_index.get(best_match, "?")

    if symbol in ['6', '9']:
        if cy > (y + h / 2):
            symbol = '6'
        else:
            symbol = '9'
            
    results_list.append((i, symbol, max_val))
    print(f"Контур {i:2d}: найден символ '{symbol}' (сходство: {max_val:.4f})")
```

```python
for i, symbol, val in results_list:
    # Берем координаты из исходного списка целевых контуров
    x, y, w, h = cv2.boundingRect(sorted_contours_target[i])
    cv2.putText(result, f"{symbol} ", (x+15, y- 10), 
                cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 0, 255), 1)
```
